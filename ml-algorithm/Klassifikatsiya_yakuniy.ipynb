# 3 Modellarni qurish va taqqoslash

Ma’lumotlarni training va test to‘plamlariga ajrating (80/20 yoki 70/30).
Quyidagi klassifikatsiya algoritmlarini qo‘llang va natijalarni solishtiring:
- K-Nearest Neighbors (KNN)
- Decision Tree
- Random Forest
- Logistic Regression

```python
X = df.drop('class', axis=1)
y = df['class']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=42, train_size=0.8)
```

```python
X_train
```

```python
# K-Nearest Neighbors (KNN)
k = []
for i in range(1,25):
  model_knn = KNeighborsClassifier(n_neighbors=i)
  model_knn.fit(X_train,y_train)
  accur_knn = accuracy_score(y_test, model_knn.predict(X_test))
  print(f"Model knn n_neighborsning {i} uchun: ", accur_knn)
  k.append(accur_knn)
```

```python
print("--"*20)
print(f"Eng yaxshi n uchun natija {k.index(max(k))+1} da {max(k)} bo'ldi")
print("--"*20)
plt.figure(figsize=(10,6))
plt.plot(range(1,25),k)
plt.grid()
plt.title("Modelning aniqligi.")
plt.show()
```

```python
sns.heatmap(confusion_matrix(y_test, model_knn.predict(X_test)), annot=True)
plt.show()
```

```python
print(classification_report(y_test, model_knn.predict(X_test)))
```

```python
# Decision Tree
model_dt = DecisionTreeClassifier(max_depth=10)
model_dt.fit(X_train, y_train)
predict_dt = model_dt.predict(X_test)
accur_dt = accuracy_score(y_test, predict_dt)
print(classification_report(y_test, predict_dt))
```

```python
sns.heatmap(confusion_matrix(y_test, model_dt.predict(X_test)), annot=True)
plt.show()
```

```python
# Random Forest

model_rf = RandomForestClassifier(n_estimators=50)
model_rf.fit(X_train, y_train)

predict_rf = model_rf.predict(X_test)
accur_rf = accuracy_score(y_test, predict_rf)
print(classification_report(y_test, predict_rf))
```

```python
sns.heatmap(confusion_matrix(y_test, model_rf.predict(X_test)), annot=True)
plt.show()
```

```python
# Logistic Regression

model_lr = LogisticRegressionCV()
model_lr.fit(X_train, y_train)

predict_lr = model_lr.predict(X_test)
accur_lr = accuracy_score(y_test, predict_lr)
print(classification_report(y_test, predict_lr))
```

# 4 Model natijalarini vizualizatsiya qilish

Har bir model uchun Confusion Matrixʼni chizib chiqing. \
Precision, Recall va F1-Score’larni bar grafigi ko‘rinishida solishtiring.

```python

# Har bir model uchun metrikalar
knn_x = [
    precision_score(y_test, model_knn.predict(X_test)),
    recall_score(y_test, model_knn.predict(X_test)),
    f1_score(y_test, model_knn.predict(X_test))
]

dt_x = [
    precision_score(y_test, model_dt.predict(X_test)),
    recall_score(y_test, model_dt.predict(X_test)),
    f1_score(y_test, model_dt.predict(X_test))
]

rf_x = [
    precision_score(y_test, model_rf.predict(X_test)),
    recall_score(y_test, model_rf.predict(X_test)),
    f1_score(y_test, model_rf.predict(X_test))
]

lr_x = [
    precision_score(y_test, model_lr.predict(X_test)),
    recall_score(y_test, model_lr.predict(X_test)),
    f1_score(y_test, model_lr.predict(X_test))
]

# X o'qi
metrics = ['Precision', 'Recall', 'F1-score']
x = np.arange(len(metrics))

# Har bir ustun kengligi
width = 0.2

plt.figure(figsize=(8,6))

plt.bar(x - 1.5*width, knn_x, width, label='KNN')
plt.bar(x - 0.5*width, dt_x, width, label='Decision Tree')
plt.bar(x + 0.5*width, rf_x, width, label='Random Forest')
plt.bar(x + 1.5*width, lr_x, width, label='Logistic Regression')

plt.xticks(x, metrics)
plt.ylim(0, 1.05)

plt.xlabel("Evaluation Metrics", fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.title("Model Performance Comparison", fontsize=14)

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```
